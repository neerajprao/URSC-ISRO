```python
# ==============================================================================
# SELF-CONTAINED DATASET INSPECTION
# ==============================================================================

import os
import glob
import numpy as np

# Find dataset files
dataset_dir = "./datasets"
npz_files = sorted(glob.glob(os.path.join(dataset_dir, "*.npz")))

print(f"Found {len(npz_files)} .npz files")

if len(npz_files) == 0:
    raise FileNotFoundError(f"No .npz files found in {dataset_dir}")

# Check first 3 files in detail
for i, fpath in enumerate(npz_files[:3]):
    print(f"\n{'='*60}")
    print(f"FILE {i+1}: {os.path.basename(fpath)}")
    print(f"{'='*60}")
    
    data = np.load(fpath, allow_pickle=True)
    
    for key in sorted(data.keys()):
        arr = data[key]
        print(f"\n  {key}:")
        print(f"    shape={arr.shape}, dtype={arr.dtype}")
        
        # Show actual content
        if arr.size <= 20:
            print(f"    values: {arr}")
        else:
            flat = arr.flatten()
            print(f"    first 5: {flat[:5]}")
            print(f"    last 5: {flat[-5:]}")
            print(f"    min={flat.min():.4f}, max={flat.max():.4f}")
    
    # Check if best_sol could be multiple layouts
    if 'best_sol' in data:
        sol = data['best_sol']
        n_elements = len(sol) // 2
        print(f"\n  >>> This layout has {n_elements} components")
        print(f"      (Our target board has 18 components)")
        
        if n_elements == 18:
            print(f"      ✓ Matches our board!")
        else:
            print(f"      ✗ Different board config")

# Summary across first 50 files
print(f"\n{'='*60}")
print("SUMMARY: Component counts across first 50 files")
print(f"{'='*60}")

comp_counts = {}
for fpath in npz_files[:50]:
    data = np.load(fpath, allow_pickle=True)
    if 'best_sol' in data:
        n = len(data['best_sol']) // 2
        comp_counts[n] = comp_counts.get(n, 0) + 1

for n, count in sorted(comp_counts.items()):
    print(f"  {n} components: {count} files")

# Check penalties
print(f"\n{'='*60}")
print("PENALTY DISTRIBUTION (first 100 files)")
print(f"{'='*60}")

penalties = []
for fpath in npz_files[:100]:
    data = np.load(fpath, allow_pickle=True)
    if 'penalty' in data:
        penalties.append(float(data['penalty'].item()))

penalties = np.array(penalties)
print(f"  Valid layouts (penalty < 1e-3): {sum(penalties < 1e-3)}/100")
print(f"  Near-valid (penalty < 1.0): {sum(penalties < 1.0)}/100")
print(f"  Median penalty: {np.median(penalties):.6f}")
print(f"  Min penalty: {penalties.min():.6f}")
print(f"  Max penalty: {penalties.max():.6f}")
```
